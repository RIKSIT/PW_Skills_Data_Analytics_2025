Question 1 : Explain the fundamental differences between DDL, DML, and DQL commands in SQL. Provide one example for each type of command.

  - DDL (Data Definition Language):
      - Used to define, modify, and delete database structures such as tables, schemas, and indexes.

      - Common commands: CREATE, ALTER, DROP, TRUNCATE

  - DML (Data Manipulation Language):
      - Used to insert, update, and delete data in tables.

     - Common commands: INSERT, UPDATE, DELETE

  - DQL (Data Query Language):
      - Used to retrieve data from database tables.

      - Common command: SELECT

  - Difference Summary:
    - DDL defines structure | DML manipulates data | DQL retrieves data.

Question 3 : Explain the difference between LIMIT and OFFSET clauses in SQL. How would you use them together to retrieve the third page of results, assuming each page has 10 records?

  - LIMIT:
     - Restricts the number of rows returned.

  - OFFSET:
    - Skips a specified number of rows before returning results.

     - Retrieve 3rd Page (10 records per page):
    - Skip first 20 records → display next 10 records.

  - Explanation:

  - Page 1 → OFFSET 0

  - Page 2 → OFFSET 10

  - Page 3 → OFFSET 20

```python
SELECT * FROM Products
LIMIT 10 OFFSET 20;
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('shop.db')
cursor = conn.cursor()
cursor.execute("""
DROP TABLE IF EXISTS Products;
""")

cursor.execute("""
CREATE TABLE Products (
    ProductID INT PRIMARY KEY,
    ProductName VARCHAR(50),
    Price INT
);
""")

conn.commit()
for i in range(1, 31):
    cursor.execute("INSERT INTO Products VALUES (?, ?, ?)",
                   (i, f'Product_{i}', i * 10))

conn.commit()
query = """
SELECT * FROM Products
LIMIT 10 OFFSET 20;
"""
pd.read_sql(query, conn)
```

Question 4 : What is a Common Table Expression (CTE) in SQL, and what are its main benefits? Provide a simple SQL example demonstrating its usage.

  - CTE (Common Table Expression):
     - A temporary named result set used inside a SQL query to improve readability and structure.

  - Benefits:

    - Improves query readability

   - Simplifies complex joins

    - Enables recursion

   - Easy debugging

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('shop.db')
cursor = conn.cursor()
cursor.execute("""
DROP TABLE IF EXISTS Products;
""")

cursor.execute("""
CREATE TABLE Products (
    ProductID INT PRIMARY KEY,
    ProductName VARCHAR(50),
    Price INT
);
""")

conn.commit()
query = """
WITH HighPriceProducts AS (
    SELECT ProductName, Price
    FROM Products
    WHERE Price > 100
)
SELECT * FROM HighPriceProducts;
"""

pd.read_sql(query, conn)
for i in range(1, 21):
    cursor.execute("INSERT INTO Products VALUES (?, ?, ?)",
                   (i, f'Product_{i}', i * 20))

conn.commit()
query = """
WITH HighPriceProducts AS (
    SELECT ProductName, Price
    FROM Products
    WHERE Price > 100
)
SELECT * FROM HighPriceProducts;
"""

pd.read_sql(query, conn)
```

Question 7 : Generate a report showing CustomerName, Email, and the TotalNumberofOrders for each customer. Include customers who have not placed any orders.

```python
query = """
SELECT
    c.CustomerName,
    c.Email,
    COUNT(o.OrderID) AS TotalNumberofOrders
FROM Customers c
LEFT JOIN Orders o ON c.CustomerID = o.CustomerID
GROUP BY c.CustomerID, c.CustomerName, c.Email
ORDER BY c.CustomerName;
"""

pd.read_sql(query, conn)
```

Question 9 : Write a SQL query that uses a Common Table Expression (CTE) and a Window Function to display the top 2 most expensive products in each CategoryName.

```python
query = """
WITH RankedProducts AS (
    SELECT
        c.CategoryName,
        p.ProductName,
        p.Price,
        ROW_NUMBER() OVER(PARTITION BY c.CategoryName ORDER BY p.Price DESC) AS rn
    FROM Products p
    JOIN Categories c ON p.CategoryID = c.CategoryID
)
SELECT CategoryName, ProductName, Price
FROM RankedProducts
WHERE rn <= 2;
"""

pd.read_sql(query, conn)

```

```python
# Install sql magic
!pip install ipython-sql sqlalchemy prettytable

# Load SQL magic
%load_ext sql

# Connect to SQLite database
%sql sqlite:///sakila.db

# 1. Top 5 customers by spending
%%sql
SELECT
    c.first_name || ' ' || c.last_name AS CustomerName,
    c.email,
    SUM(p.amount) AS TotalSpent
FROM customer c
JOIN payment p ON c.customer_id = p.customer_id
GROUP BY c.customer_id, c.first_name, c.last_name, c.email
ORDER BY TotalSpent DESC
LIMIT 5;


# 2. Top 3 movie categories by rentals
%%sql
SELECT
    cat.name AS CategoryName,
    COUNT(*) AS RentalCount
FROM rental r
JOIN inventory i ON r.inventory_id = i.inventory_id
JOIN film_category fc ON i.film_id = fc.film_id
JOIN category cat ON fc.category_id = cat.category_id
GROUP BY cat.name
ORDER BY RentalCount DESC
LIMIT 3;


# 3. Films available and never rented per store
%%sql
SELECT
    s.store_id,
    COUNT(i.inventory_id) AS TotalFilms,
    SUM(CASE WHEN r.rental_id IS NULL THEN 1 ELSE 0 END) AS NeverRented
FROM store s
JOIN inventory i ON s.store_id = i.store_id
LEFT JOIN rental r ON i.inventory_id = r.inventory_id
GROUP BY s.store_id;


# 4. Monthly revenue for 2023
%%sql
SELECT
    strftime('%m', payment_date) AS Month,
    SUM(amount) AS TotalRevenue
FROM payment
WHERE strftime('%Y', payment_date) = '2023'
GROUP BY Month
ORDER BY Month;


# 5. Customers rented more than 10 times in last 6 months
%%sql
SELECT
    c.customer_id,
    c.first_name || ' ' || c.last_name AS CustomerName,
    COUNT(r.rental_id) AS RentalCount
FROM customer c
JOIN rental r ON c.customer_id = r.customer_id
WHERE r.rental_date >= date('now','-6 months')
GROUP BY c.customer_id, c.first_name, c.last_name
HAVING COUNT(r.rental_id) > 10;
```
