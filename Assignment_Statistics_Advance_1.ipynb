Question 1: What is a random variable in probability theory?


In probability theory, a random variable is a variable whose possible values are numerical outcomes of a random phenomenon.

Formally:
A random variable is a function that assigns a real number to each outcome in the sample space of a random experiment.

Types of Random Variables

Discrete Random Variable

Takes on a countable number of distinct values (e.g., number of heads in 3 coin tosses).

Example:
X= number of students present in class today.

Continuous Random Variable

Takes on infinitely many possible values within an interval (e.g., height, time, temperature).

Example:
Y= the time (in seconds) it takes for a website to load.

Question 2: What are the types of random variables?


In probability theory, random variables are broadly classified into two main types, based on the nature of their possible values:


1. Discrete Random Variable

Takes countable values (finite or countably infinite).

Usually represents counts or distinct outcomes.

Probability distribution is described using a Probability Mass Function (PMF).

Examples:

Number of heads in 5 coin tosses (X=0,1,2,3,4,5).

Number of students absent in a class.

Result of rolling a die (1,2,3,4,5,6).



2. Continuous Random Variable

Takes uncountably infinite values (over an interval of real numbers).

Typically represents measurements.

Probability distribution is described using a Probability Density Function (PDF).

The probability of the variable taking any exact single value is 0; we calculate probabilities over intervals.

Examples:

Height or weight of a person.

Time taken to run a race.

Temperature in a city on a given day.

Question 4: What is a binomial distribution, and how is it used in probability?


The binomial distribution is a discrete probability distribution that describes the number of successes in a fixed number of independent Bernoulli trials (yes/no experiments), each with the same probability of success.


Definition

A random variable
X follows a Binomial Distribution if:

The experiment consists of
n independent trials.

Each trial has two possible outcomes: success (probability p) or failure (probability q=1−p).


X counts the number of successes in the n trials.

Question 6: What is the Central Limit Theorem (CLT), and why is it critical in statistics?


The Central Limit Theorem is a fundamental result in statistics that describes the behavior of sample means.


Key Points

Works even if the population is not normally distributed.

The larger the sample size 𝑛
n, the closer the sample mean distribution is to normal.


Why is CLT Critical?

Foundation for Inferential Statistics:

Allows us to make probability statements about sample means.

Justifies the use of Z-tests, T-tests, and confidence intervals.

Simplifies Analysis:

We can assume sample means are approximately normal, even if the underlying population is not.

Supports Large-Sample Approximations:

For large n, we can use normal tables for calculating probabilities.


Example

Population: Skewed distribution of daily sales (not normal).

Take 50-day samples, compute sample mean each time.

According to CLT, the distribution of these sample means will approximate a normal distribution.

Question 7: What is the significance of confidence intervals in statistical analysis?

A confidence interval is a range of values used to estimate an unknown population parameter (like the mean or proportion) based on sample data.

It gives us an interval within which we are “confident” that the true population parameter lies.

Expressed as:

CI=point estimate±margin of error

CI=point estimate±margin of error

Significance of Confidence Intervals

Provides a Range, Not a Single Estimate:

Unlike a point estimate (e.g., sample mean), CI accounts for sampling variability.

Gives a sense of precision of the estimate.

Measures Uncertainty:

Wider CI → more uncertainty in the estimate.

Narrower CI → more precise estimate.

Supports Decision Making:

Helps in hypothesis testing. For example, if a CI for a difference in means does not include 0, we can infer a significant difference.

Communicates Reliability:

A 95% CI means that if we repeat the sampling many times, approximately 95% of those intervals will contain the true parameter.

Question 8: What is the concept of expected value in a probability distribution?

The expected value (also called the mean) of a random variable is the long-run average value it would take if an experiment is repeated many times.

It provides a measure of the “center” or “typical” value of a probability distribution.


Interpretation

Expected value is not necessarily a value the random variable can actually take.

It represents the average outcome in the long run.


Significance

Decision Making:

Used in games of chance, insurance, and finance to calculate average expected payoff.

Measure of Center:

Helps summarize a probability distribution by a single representative value.

Foundation for Variance and Other Moments:

Expected value is used in calculating variance, standard deviation, and higher moments.

```python
"""Question 9: Write a Python program to generate 1000 random numbers from a normal
distribution with mean = 50 and standard deviation = 5. Compute its mean and standard
deviation using NumPy, and draw a histogram to visualize the distribution."""


import numpy as np
import matplotlib.pyplot as plt

# Parameters
mean = 50
std_dev = 5
n_samples = 1000

# Generate 1000 random numbers from normal distribution
data = np.random.normal(loc=mean, scale=std_dev, size=n_samples)

# Compute mean and standard deviation
calculated_mean = np.mean(data)
calculated_std = np.std(data)

print(f"Calculated Mean: {calculated_mean:.2f}")
print(f"Calculated Standard Deviation: {calculated_std:.2f}")

# Draw histogram
plt.figure(figsize=(10,6))
plt.hist(data, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

```

Step 1: Applying CLT to Estimate Average Sales

Problem Setup:

We have daily sales data (daily_sales) for 20 days (for simplicity, assume it’s a representative sample).

The company wants to estimate the average daily sales for the population (all days).

Central Limit Theorem Application:

CLT states that the sampling distribution of the sample mean is approximately normal for sufficiently large sample sizes.
Confidence Interval (CI):

For a 95% confidence interval, we use the Z-value ≈ 1.96 (for large n)

```python
#Step 2: Python Code to Compute Mean and 95% CI

import numpy as np
from scipy import stats

# Daily sales data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Convert to NumPy array
data = np.array(daily_sales)

# Sample statistics
n = len(data)
mean_sales = np.mean(data)
std_sales = np.std(data, ddof=1)  # Sample standard deviation

# Standard error
se = std_sales / np.sqrt(n)

# 95% confidence interval using t-distribution
confidence_level = 0.95
t_value = stats.t.ppf((1 + confidence_level) / 2, df=n-1)
margin_of_error = t_value * se

ci_lower = mean_sales - margin_of_error
ci_upper = mean_sales + margin_of_error

print(f"Sample Mean Sales: {mean_sales:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
```
