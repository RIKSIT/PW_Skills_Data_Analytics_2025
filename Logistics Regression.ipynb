Purpose of Simple Linear Regression

The main purposes are:

Prediction:
To predict the value of the dependent variable (𝑌) for a given value of the independent variable (X).
Example: Predicting house prices based on square footage.

Relationship Analysis:
To determine if and how strongly two variables are related — e.g., whether an increase in advertising budget leads to higher sales.

Trend Estimation:
To understand general trends and patterns in data (e.g., how temperature affects electricity consumption).


Question 2: What are the key assumptions of Simple Linear Regression?

The key assumptions of Simple Linear Regression (SLR) ensure that the model’s estimates and predictions are valid and reliable.

Here are the main assumptions explained clearly:

🔹 1. Linearity

The relationship between the independent variable (X) and the dependent variable (Y) is linear.

This means changes in 
X lead to proportional changes in 𝑌


Check: Scatter plot of X and Y should show a straight-line trend.

🔹 2. Independence of Errors

The residuals (errors) are independent of each other.

There should be no correlation between the errors of different observations.

Check: Use Durbin-Watson test (especially for time series data).

🔹 3. Homoscedasticity (Constant Variance of Errors)

The variance of the residuals is constant across all levels of X.

In other words, the spread of residuals should be roughly the same for all fitted values.
Check: Residual vs. fitted value plot should show no funnel shape.

4. Normality of Errors

The residuals (errors) should be normally distributed with mean 0.

This is especially important for hypothesis testing and confidence intervals.

Check: Histogram or Q-Q plot of residuals should look approximately normal.

5. No Perfect Multicollinearity

In Simple Linear Regression, there’s only one independent variable, so this assumption automatically holds.

(Relevant mainly for Multiple Linear Regression.)

6. No Measurement Error in X

The independent variable (X) is assumed to be measured without error.

Measurement errors in X can lead to biased estimates of the slope.


Question 3: Write the mathematical equation for a simple linear regression model and

The mathematical equation for a Simple Linear Regression (SLR) model is:

𝑌=𝛽0+𝛽1𝑋+𝜀

Y=Dependent variable (Response variable)
X=Independent variable (Predictor variable)
β0=Intercept (Constant term)
β1=Slope (Regression coefficient)
ε=Error term (Residual)

Example

Suppose we are predicting monthly sales (Y) based on advertising spending (X):

Sales=50+5×Advertising+𝜀

𝛽0=50: When advertising = 0, expected sales = 50 units.
β1=5: For every additional $1 spent on advertising, sales increase by 5 units (on average)
ε: Accounts for other random influences on sales.

Question 6: What is Logistic Regression? How does it differ from Linear Regression?


Logistic Regression is a statistical and machine learning technique used for classification problems, particularly when the dependent variable (Y) is binary — meaning it has only two possible outcomes (e.g., 0/1, Yes/No, Pass/Fail).

It models the probability that 
Y belongs to a particular category (usually coded as 1) based on one or more independent variables (X).

Purpose

To classify observations into categories (e.g., “spam” vs “not spam”).

To estimate probabilities of outcomes.

Logistic Regression is used when the outcome is categorical, modeling the probability of an event.

Linear Regression is used when the outcome is continuous, modeling a straight-line relationship between variables.

Question 7: Name and briefly describe three common evaluation metrics for regression
models.


1. Mean Absolute Error (MAE)

MAE measures the average magnitude of the errors between predicted and actual values, without considering their direction.
It shows how far, on average, the predictions are from the true values.
A lower MAE indicates that the model’s predictions are more accurate.

2. Mean Squared Error (MSE)

MSE measures the average of the squared differences between predicted and actual values.
By squaring the errors, it gives more weight to larger errors, making it sensitive to outliers.
A smaller MSE value indicates better model performance.

3. Coefficient of Determination (R²)

R² indicates the proportion of the variance in the dependent variable that is explained by the independent variable(s) in the model.
It shows how well the regression line fits the data.
An R² value closer to 1 means the model explains most of the variability in the outcome variable.


Question 8: What is the purpose of the R-squared metric in regression analysis?

The R-squared (R²) metric — also known as the coefficient of determination — measures how well a regression model explains the variability of the dependent variable using the independent variable(s).

🔹 Main Purpose

The purpose of R² is to evaluate the goodness of fit of a regression model — that is, how well the model’s predictions match the actual data.

🔹 Interpretation

R² represents the proportion of variance in the dependent variable that can be explained by the independent variable(s).

Its value ranges between 0 and 1:

R² = 1 → Perfect fit (model explains all variability in the data).

R² = 0 → Model explains none of the variability (no relationship).

Higher R² → Better explanatory power of the model.


```python
"""Question 9: Write Python code to fit a simple linear regression model using scikit-learn
and print the slope and intercept."""

# Import necessary libraries
from sklearn.linear_model import LinearRegression
import numpy as np

# Example data
# X: independent variable (must be 2D for scikit-learn)
# Y: dependent variable
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2, 4, 5, 4, 5])

# Create a Linear Regression model
model = LinearRegression()

# Fit the model to the data
model.fit(X, Y)

# Get the slope (coefficient) and intercept
slope = model.coef_[0]
intercept = model.intercept_

# Print results
print(f"Slope (β1): {slope}")
print(f"Intercept (β0): {intercept}")
```
