Question 1: What is Information Gain, and how is it used in Decision Trees?
  - Definition: Information Gain (IG) measures the reduction in uncertainty (entropy) after a dataset is split based on a particular attribute.

      - Formula:
      - IG(S,A)=Entropy(S)−v∈Values(A)∑​∣S∣∣Sv​∣​×Entropy(Sv​)
  - Purpose: It helps determine which feature should be chosen for splitting at each node in the decision tree.

  - Working: Higher Information Gain means the feature provides more information to classify the data correctly.

  - Usage in Decision Tree: The feature with the highest Information Gain is selected as the root or next internal node.

  - Example: If splitting by “Outlook” gives the highest IG, it becomes the first split in the tree.

Question 2: What is the difference between Gini Impurity and Entropy?

  - Gini Impurity measures how often a randomly chosen element would be incorrectly labeled if it were randomly labeled according to the distribution of labels in the dataset.

  - Entropy measures the level of disorder or uncertainty in the dataset.

      - Formula:
    - Gini Impurity = 1−∑pi2​
    - Entropy = -∑pi​log2​(pi​)

      - Value Range:

  - Gini ranges from 0 (pure) to 0.5 (most impure for binary class).

  - Entropy ranges from 0 (pure) to 1 (maximum disorder for binary class).

      - Interpretation:

  - Gini focuses on probability of incorrect classification.

  - Entropy focuses on information or surprise in the data.

      - Computation Speed:

  - Gini is faster to compute since it avoids logarithmic calculations.

  - Entropy is slightly slower because it uses log operations.

      - Algorithm Usage:

  - Gini is used by the CART (Classification and Regression Tree) algorithm.

  - Entropy is used by ID3 and C4.5 algorithms.

      - Preference:

  - Gini is preferred when computation efficiency matters.

  - Entropy is preferred when interpretability or information theory perspective is important.

Question 3: What is Pre-Pruning in Decision Trees?


  - Definition: Pre-pruning (also called early stopping) stops the tree growth early before it perfectly fits the training data.

  - Goal: To prevent overfitting by setting limits while building the tree.

  - Common Criteria:

  - Maximum tree depth (max_depth)

  - Minimum samples required to split (min_samples_split)

  - Minimum samples per leaf (min_samples_leaf)

  - Minimum Information Gain threshold

  - Advantage: Saves computation time and improves generalization to unseen data.

  - Disadvantage: May underfit if the tree stops growing too early.

Question 4: Python Program — Decision Tree Classifier using Gini Impurity

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# Load dataset
data = load_iris()
X = data.data
y = data.target

# Train Decision Tree with Gini criterion
clf = DecisionTreeClassifier(criterion='gini', random_state=42)
clf.fit(X, y)

# Print feature importances
feature_importances = pd.Series(clf.feature_importances_, index=data.feature_names)
print("Feature Importances:\n", feature_importances)
```

Question 5: What is a Support Vector Machine (SVM)?


  - Definition: SVM is a supervised machine learning algorithm used for classification and regression tasks.

  - Concept: It finds an optimal hyperplane that separates data points of different classes with maximum margin.

  - Support Vectors: The data points closest to the hyperplane that influence its position.

  - Goal: Maximize the margin between classes to improve model robustness.

  - Advantages: Works well for high-dimensional data and clear class separation.

  - Limitation: Performance drops when data is heavily overlapping or noisy.

Question 6: What is the Kernel Trick in SVM?

  - Definition: The Kernel Trick allows SVM to perform classification in a higher-dimensional space without explicitly transforming the data.

  - Purpose: Helps in separating non-linearly separable data by mapping it into a new feature space.

  - Common Kernels:

  - Linear Kernel

  - Polynomial Kernel

  - Radial Basis Function (RBF) Kernel

  - Sigmoid Kernel

  - Advantage: Saves computation time and memory since it avoids explicitly computing coordinates in the higher dimension.

  - Example: The RBF kernel maps circularly separated data into a linearly separable space.

Question 7: Python Program — SVM Classifiers with Linear and RBF Kernels

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load dataset
wine = datasets.load_wine()
X_train, X_test, y_train, y_test = train_test_split(wine.data, wine.target, test_size=0.3, random_state=42)

# Linear Kernel
svm_linear = SVC(kernel='linear')
svm_linear.fit(X_train, y_train)
linear_acc = accuracy_score(y_test, svm_linear.predict(X_test))

# RBF Kernel
svm_rbf = SVC(kernel='rbf')
svm_rbf.fit(X_train, y_train)
rbf_acc = accuracy_score(y_test, svm_rbf.predict(X_test))

print("Linear Kernel Accuracy:", linear_acc)
print("RBF Kernel Accuracy:", rbf_acc)
```

Question 8: What is the Naïve Bayes classifier, and why is it called "Naïve"?

Answer:

  - Definition: Naïve Bayes is a probabilistic classifier based on Bayes’ Theorem that assumes independence between predictors.

  - Formula:
      - P(C∣X)=P(X)P(X∣C)×P(C)
  - Naïve Assumption: It assumes all features contribute independently to the outcome — hence called “Naïve.”

  - Use Case: Text classification, spam filtering, sentiment analysis.

  - Advantages: Simple, fast, works well with small datasets.

  - Disadvantage: Performs poorly when features are highly correlated.​

Question 9: Explain the differences between Gaussian, Multinomial, and Bernoulli Naïve Bayes.

  - Data Type:

  - Gaussian Naïve Bayes is used for continuous (real-valued) data.

  - Multinomial Naïve Bayes is used for discrete count data.

  - Bernoulli Naïve Bayes is used for binary or Boolean data (0s and 1s).

    - Underlying Distribution:

  - Gaussian NB assumes that features follow a normal (bell-shaped) distribution.

  - Multinomial NB assumes that features represent event counts following a multinomial distribution.

  - Bernoulli NB assumes binary presence/absence of features.

    - Example Use Cases:

  - Gaussian NB: Predicting diseases using continuous medical data.

  - Multinomial NB: Text classification based on word frequencies.

  - Bernoulli NB: Spam detection based on presence or absence of certain words.

    - Feature Representation:

  - Gaussian NB works with real numbers like age, height, or temperature.

  - Multinomial NB works with integer counts (e.g., word count = 3).

  - Bernoulli NB works with binary indicators (e.g., word present = 1, absent = 0).

    -Suitability:

  - Gaussian NB suits continuous numeric features.

  - Multinomial NB suits count-based text data.

  - Bernoulli NB suits binary or boolean data representation.

    - Performance Aspect:

  - Gaussian performs best when data is normally distributed.

  - Multinomial performs best with frequency-based features.

  - Bernoulli performs best when features are binary and sparse.

Question 10: Python Program — Gaussian Naïve Bayes on Breast Cancer Dataset

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load dataset
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.3, random_state=42)

# Train Gaussian Naive Bayes model
gnb = GaussianNB()
gnb.fit(X_train, y_train)

# Predictions and accuracy
y_pred = gnb.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Gaussian Naive Bayes Accuracy:", accuracy)
```
