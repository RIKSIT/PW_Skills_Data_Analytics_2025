Question 1: What is the difference between 'Paste' and 'Paste Special' in Excel? Briefly explain with examples.

  - Paste:

     - The Paste option pastes everything exactly as copied, including values, formulas, formatting, and comments.

      - It is used when we want a complete duplicate of the copied data.

     - Example: If you copy a cell containing the formula =A1+B1, pasting it will paste the formula itself, not just the result.

  - Paste Special:

      - Paste Special allows selective pasting, such as only values, only formulas, only formats, transpose, operations, etc.

     - It provides greater control over pasted data.

      - Example: If a formula result is 50, using Paste Special → Values pastes only 50, not the formula.

Question 2: Describe the functions and usefulness of 'Freeze Panes' and 'Split Panes' in Excel.

  - Freeze Panes:

      - Keeps selected rows or columns visible while scrolling.

      - Useful when working with large datasets.

     - Example: Freezing the header row allows column titles to remain visible while scrolling down.

  - Split Panes:

      - Divides the worksheet window into multiple scrollable sections.

     - Helps compare data from different parts of the same worksheet.

     - Example: Viewing top and bottom data simultaneously.
  - Usefulness:

     - Freeze Panes improves readability.

      - Split Panes improves data comparison and navigation.

Question 3: Explain the difference between inserting a new row and inserting a new column in Excel. Can you insert multiple rows or columns at once?

  - Insert New Row:

  - Adds a new horizontal row.

  - Shifts existing rows downward.

  - Used when adding new records.

  - Insert New Column:

  - Adds a new vertical column.

  - Shifts existing columns to the right.

  - Used when adding new data fields.

  - Inserting Multiple Rows or Columns:

  - Yes, select multiple rows or columns, right-click, and choose Insert.
  - Example:
  - Selecting 3 rows and clicking Insert will insert 3 new rows at once.

Question 4: What are logical functions in Excel? Provide examples of at least two logical functions and their applications.

  - Logical functions perform logical tests and return results based on conditions (TRUE or FALSE).

     - Examples:

     - IF Function:

     - Formula: =IF(A1>=50,"Pass","Fail")

     - Application: Determines pass or fail status based on marks.

  - AND Function:

      - Formula: =AND(A1>=50, B1>=50)

     - Application: Checks multiple conditions together.

  - Usefulness:
    - Logical functions help in decision-making, automation, and data validation.

Question 5: Discuss the purpose of 'XLOOKUP' and how it differs from the traditional 'VLOOKUP' function.

  - Purpose of XLOOKUP:

      - Finds values in a range or table and returns matching results.

      - Works both vertically and horizontally.

  - Differences:

      - Feature	VLOOKUP	XLOOKUP
      - Direction	Only left-to-right	Both directions
      - Column Index	Required	Not required
      - Error Handling	Complex	Built-in
      - Flexibility	Limited	High

  - Example:
    - =XLOOKUP(A2,A:A,B:B) finds value in column A and returns matching value from B.

Question 10: You're working with a dataset that contains missing values. As a Data Scientist, explain how you'd detect and handle missing data using Excel. Mention tools like:

● Go To Special
● ISBLANK
● COUNTBLANK

  - Detect Missing Data:

      - 1. Go To Special:

         - Home → Find & Select → Go To Special → Blanks

         - Highlights all blank cells instantly.

     - 2. ISBLANK Function:
       - Formula: =ISBLANK(A1)

       - Returns TRUE if blank.

      - 3. COUNTBLANK Function:
        - Formula: =COUNTBLANK(A1:A50)

        - Counts total missing cells.

        - Handling Missing Data:

        - Delete rows

        - Fill using mean/median

        - Replace with suitable default values
