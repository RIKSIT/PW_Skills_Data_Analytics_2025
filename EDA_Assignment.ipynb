## Question 1: Difference between AI, ML, DL, and Data Science

**Answer:**

**Artificial Intelligence (AI):**
- Scope: Broad field focused on creating machines/systems that can perform tasks that normally require human intelligence.
- Techniques: Rule-based systems, search algorithms, planning, knowledge representation, optimization, ML methods.
- Applications: Expert systems, robotics, game playing, natural language processing, decision support.

**Machine Learning (ML):**
- Scope: Subset of AI focused on algorithms that learn from data to make predictions or decisions without explicit programming.
- Techniques: Supervised learning (regression, classification), unsupervised learning (clustering, dimensionality reduction), reinforcement learning.
- Applications: Spam detection, recommendation systems, fraud detection, predictive maintenance.

**Deep Learning (DL):**
- Scope: Subset of ML that uses multi-layer neural networks (deep neural networks) to model complex patterns.
- Techniques: Convolutional Neural Networks (CNNs), Recurrent Neural Networks (RNNs), Transformers, autoencoders.
- Applications: Image recognition, speech recognition, language models, machine translation.

**Data Science:**
- Scope: Interdisciplinary field that combines domain knowledge, statistics, data engineering, and ML to extract insights from data.
- Techniques: Data cleaning, exploratory data analysis (EDA), visualization, statistical inference, ML modeling, deployment.
- Applications: Business analytics, scientific research, decision-making, A/B testing.

**Comparison summary:**
- AI is the broad goal (intelligent systems). ML is the practice of learning from data within AI. DL is a powerful technique within ML specialized for large data and complex patterns. Data Science is an end-to-end discipline that uses ML/DL among other methods to solve data problems and communicate insights.


## Question 2: Overfitting and Underfitting — Detection and Prevention

**Answer:**

**Overfitting:**
- Definition: A model fits the training data too well, capturing noise and specific patterns that do not generalize to new data. Overfitted models have low training error but high validation/test error.
- Detection: Large gap between training and validation performance (e.g., training accuracy high, validation accuracy low). Learning curves show training error decreasing while validation error increases.
- Prevention / Remedies:
  - Get more data (if possible).
  - Regularization: L1 (lasso) or L2 (ridge) penalties, dropout for neural networks.
  - Simplify model: reduce number of features, prune decision trees, reduce network size.
  - Cross-validation: use k-fold CV to get robust performance estimates and tune hyperparameters.
  - Early stopping (stop training when validation error stops improving).

**Underfitting:**
- Definition: A model is too simple to capture the underlying structure in the data; both training and validation errors are high.
- Detection: Both training and validation performance are poor and close to each other.
- Remedies:
  - Increase model complexity (use more expressive models).
  - Add informative features or use feature engineering.
  - Reduce strong regularization.
  - Train longer or use better architectures (for DL).

**Bias–Variance Tradeoff:**
- Bias: Error due to erroneous assumptions in the learning algorithm (underfitting).
- Variance: Error due to sensitivity to small fluctuations in the training set (overfitting).
- Goal: Find a balance where total error (bias^2 + variance + irreducible error) is minimized. Cross-validation and validation curves help find this balance.

**Cross-validation & Model Selection:**
- Use k-fold cross-validation to estimate model generalization.
- Use the validation set or nested CV for hyperparameter tuning to avoid optimistic bias.


## Question 3: Handling Missing Values — Methods with Examples

**Answer:**

1. **Deletion (Listwise or Pairwise):**
   - Listwise deletion: Remove rows with missing values. Simple but can bias results if missingness is not completely at random and reduces dataset size.
   - Pairwise deletion: Use available data per analysis; more complicated to implement consistently.
   - Example: `df.dropna()` in pandas.

2. **Imputation (Mean/Median/Mode):**
   - Replace missing numeric values with mean or median (median for skewed distributions).
   - Replace categorical missing values with mode.
   - Example:
     ```python
     df['age'].fillna(df['age'].median(), inplace=True)
     df['city'].fillna(df['city'].mode()[0], inplace=True)
     ```
   - Pros: Simple, preserves dataset size. Cons: Underestimates variability and can bias correlations.

3. **Predictive Modeling (KNN, Regression Imputation, or MICE):**
   - Use other features to predict missing values. KNN imputes based on values from nearest neighbors.
   - Multiple Imputation by Chained Equations (MICE) fits models for each variable with missing values iteratively to account for uncertainty.
   - Example (sketch):
     ```python
     from sklearn.impute import KNNImputer
     imputer = KNNImputer(n_neighbors=5)
     df_imputed = pd.DataFrame(imputer.fit_transform(df_num), columns=df_num.columns)
     ```

4. **Indicator for Missingness:**
   - Add a boolean column indicating missingness; useful when missingness itself is informative.
   - Example: `df['age_missing'] = df['age'].isna().astype(int)`

**Choosing a method:** depends on missingness mechanism (MCAR, MAR, MNAR), amount of missing data, and analysis goals.


## Question 4: Imbalanced Dataset — Definition and Two Techniques

**Answer:**

**Definition:** Imbalanced dataset is one where class distribution is skewed — one or more classes have many more examples than others. This can cause models to be biased toward majority class.

**Technique 1 — Resampling:**
- **Oversampling:** Increase minority class examples (random oversampling or SMOTE — Synthetic Minority Over-sampling Technique).
  - SMOTE creates synthetic samples in feature space by interpolating between minority class neighbors.
  - Example: `imblearn.over_sampling.SMOTE()`.
- **Undersampling:** Reduce majority class examples (random undersampling or informed undersampling).
  - Risk: losing potentially useful data.

**Technique 2 — Algorithmic / Cost-sensitive Methods:**
- **Class weights:** In models like `LogisticRegression`, `RandomForestClassifier`, or many XGBoost/LightGBM implementations, pass `class_weight='balanced'` or custom weights so misclassifying minority class is penalized more.
- **Threshold adjustment and metrics:** Use precision-recall, F1-score, AUC-PR instead of accuracy. Adjust probability threshold to favor minority recall/precision tradeoff.

**Practical workflow:**
1. Inspect class distribution and baseline metrics.
2. Try class weights or resampling and evaluate using cross-validation and relevant metrics (precision, recall, F1, PR-AUC).
3. For high-dimensional data, experiment carefully with SMOTE variants (SMOTEN, SMOTENC for categorical features).


## Question 5: Feature Scaling — Importance and Comparison (Min-Max vs Standardization)

**Answer:**

**Why scaling matters:**
- Many algorithms rely on distances (KNN, K-means) or assume features are on similar scales (SVM, logistic regression with regularization). If features are on different scales, the algorithm may be dominated by features with larger numeric ranges.
- Gradient-based optimizers converge faster when features are scaled.

**Min-Max Scaling (Normalization):**
- Formula: `X_scaled = (X - X_min) / (X_max - X_min)` → maps to range [0, 1] (or custom range).
- Pros: Preserves shape of original distribution; bounds are fixed.
- Cons: Sensitive to outliers; if new data has values outside training range, transforms can go out of [0,1].

**Standardization (Z-score):**
- Formula: `X_scaled = (X - mean) / std` → results in zero mean and unit variance.
- Pros: Less sensitive to outliers than min-max; good for algorithms that assume centered data.
- Cons: Does not bound values to a fixed interval.

**When to use which:**
- Use **Min-Max** when features are bounded and you need values in a specific range (e.g., neural network inputs with activation ranges).
- Use **Standardization** for most ML algorithms, especially those assuming Gaussian-like distributions or using regularization.

**Impact on algorithms:**
- Distance-based models (KNN, K-means): scaling is critical.
- Regularized models (Ridge, Lasso, Logistic Regression): scaling matters because penalty depends on magnitude of coefficients.
- Tree-based models (Decision Trees, Random Forests): not sensitive to scaling.


## Question 6: Label Encoding vs One-Hot Encoding

**Answer:**

**Label Encoding:**
- Assigns integer values to categories (e.g., `Red`→0, `Green`→1, `Blue`→2).
- Use when categorical variable is ordinal (has natural order) because integers preserve order for many models.
- Risk: For nominal categories, integers introduce an artificial ordinal relationship which can mislead models.

**One-Hot Encoding:**
- Creates binary indicator columns for each category (e.g., `is_red`, `is_green`, `is_blue`).
- Use for nominal (unordered) categories to avoid introducing artificial order.
- Pros: Works well for linear models and many ML algorithms. Cons: High dimensionality when many categories (curse of dimensionality).

**When to prefer which:**
- Ordinal categories (e.g., `low`, `medium`, `high`): label encode or map to meaningful numeric scale.
- Nominal categories with few levels: one-hot encode.
- High-cardinality nominal categories: consider target encoding, hashing encoding, or embedding techniques (for deep learning).


## Question 7: Google Play Store Dataset — Analysis of Categories vs Ratings

**Answer (approach + code template + interpretation):**

**Approach:**
1. Load dataset into a DataFrame.
2. Clean ratings column (convert to numeric, handle missing values/outliers).
3. Group by `Category` and compute mean and median rating, sample size, and distribution.
4. Visualize with barplots and boxplots to capture central tendency and spread.

**Code template (runnable):**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Replace the URL with local path or raw GitHub URL
url = 'GOOGLE_PLAY_STORE_CSV'
df = pd.read_csv(url)

# Basic cleaning
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
cat_stats = df.groupby('Category')['Rating'].agg(['mean','median','count','std']).sort_values('mean', ascending=False)
print(cat_stats)

# Barplot (average rating)
plt.figure(figsize=(10,6))
sns.barplot(data=cat_stats.reset_index(), x='mean', y='Category')
plt.title('Average Rating by Category')
plt.show()

# Boxplot to show spread
plt.figure(figsize=(12,8))
sns.boxplot(data=df, x='Rating', y='Category')
plt.title('Rating Distribution by Category')
plt.show()
```

**Typical interpretation (what to look for):**
- Categories with the highest average ratings are often niche categories with fewer apps (small-sample effect), or apps that are utility-focused and well-maintained.
- Categories with lowest average ratings may be those with many free apps, ad-supported apps, or categories with higher user expectation (e.g., social/communication).
- Always consider sample size: categories with few apps can have volatile averages; show `count` alongside mean.


## Question 8: Titanic Dataset

**Answer (approach + code template + expected interpretation):**

**(a) Survival rates by passenger class (Pclass):**

**Approach:**
1. Load the Titanic dataset.
2. Compute survival rate per `Pclass` (`survived.mean()` grouped by `Pclass`).
3. Visualize using bar plots.

**Code template:**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = 'TITANIC_CSV'
df = pd.read_csv(url)

pclass_surv = df.groupby('Pclass')['Survived'].mean().reset_index().sort_values('Survived', ascending=False)
print(pclass_surv)

sns.barplot(data=pclass_surv, x='Pclass', y='Survived')
plt.title('Survival Rate by Pclass')
plt.ylim(0,1)
plt.show()
```

**Typical finding:** Historically, **1st class** passengers had the highest survival rate due to better access to lifeboats and evacuation priority, while 3rd class (steerage) had the lowest.

**(b) Age effect — children vs adults:**

**Approach:**
1. Create a binary `is_child` column: `df['is_child'] = df['Age'] < 18`.
2. Compare survival rates between children and adults overall and within Pclass and Sex groups.

**Code template:**

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['is_child'] = (df['Age'] < 18).astype(int)
child_surv = df.groupby('is_child')['Survived'].mean().reset_index()
print(child_surv)

# Compare within Pclass
pd.crosstab(df['Pclass'], df['is_child']).head()

sns.barplot(data=df, x='is_child', y='Survived')
plt.xticks([0,1], ['Adult','Child'])
plt.title('Survival Rate: Children vs Adults')
plt.ylim(0,1)
plt.show()
```

**Typical observation:** Children often had a higher chance of survival overall, particularly among certain classes and when crew prioritized 'women and children'. But results can vary by class and gender; analyze within groups.

## Question 10: HR Analytics Dataset — Attrition Analysis

**Answer (approach + code template + interpretation):**

**(a) Factors correlated with employee attrition:**

**Approach:**
1. Load `hr_analytics.csv`.
2. Compute correlation (for numeric features) with `Attrition` encoded as 0/1. Use point-biserial for binary/numeric relations and chi-square for categorical.
3. Use visualizations (boxplots, violin plots, countplots) for features like `Satisfaction`, `OverTime`, `MonthlyIncome`, `NumProjects`.
4. Fit a simple logistic regression and inspect coefficients or use SHAP for model-agnostic feature importance.

**Code template:**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

url = 'HR_ANALYTICS_CSV'
df = pd.read_csv(url)

# Example preprocessing
le = LabelEncoder()
df['Attrition_flag'] = le.fit_transform(df['Attrition'])  # assuming 'Yes'/'No'

# Correlations
corrs = df.select_dtypes(include='number').corr()['Attrition_flag'].sort_values()
print(corrs)

# Visuals
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')
plt.title('Income vs Attrition')
plt.show()

sns.countplot(data=df, x='OverTime', hue='Attrition')
plt.title('OverTime vs Attrition')
plt.show()
```

**Interpretation guidance:**
