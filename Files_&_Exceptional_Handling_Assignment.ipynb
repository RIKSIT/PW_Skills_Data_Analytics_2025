1.What is the difference between interpreted and compiled languages?

The main difference between interpreted and compiled languages lies in how their code is executed by the computer. Here's a clear breakdown:

Compiled Languages

The source code (what you write) is translated all at once into machine code (binary instructions the CPU understands) by a compiler.

This produces an executable file that can run directly on the computer without needing the original source code or the compiler.

Because it's pre-translated, compiled programs generally run faster.

Examples: C, C++, Rust, Go.

Interpreted Languages

The source code is executed line-by-line or statement-by-statement by an interpreter at runtime.

There is no separate executable file; the interpreter reads the source code each time you run the program.

This usually makes interpreted programs slower than compiled ones.

Examples: Python, JavaScript, Ruby, PHP.













2.What is exception handling in Python?


Exception handling in Python is a way to manage errors that occur during the execution of a program, so your program doesn’t crash abruptly. Instead, it allows you to catch these errors (called exceptions) and respond to them gracefully.

How it works:

When Python encounters an error, it raises an exception.

You can catch this exception using a try block paired with one or more except blocks.

The code inside the try block is executed normally.

If an error happens inside the try block, Python stops executing that block and looks for an except block that matches the type of exception.

If a matching except block is found, the code inside it runs.

If no matching except block is found, the program crashes with an error.


try:
    # Code that might cause an exception
    result = 10 / 0
except ZeroDivisionError:
    # Code to run if ZeroDivisionError occurs
    print("You can't divide by zero!")

9.What is the difference between multithreading and multiprocessing?

Multithreading and multiprocessing are both ways to run code concurrently in Python (or other languages), but they work quite differently under the hood and are suited for different tasks.


-Multithreading

Involves multiple threads running within the same process.

Threads share the same memory space, so they can easily share data.

Because threads share memory, you need to manage access carefully to avoid conflicts (using locks, semaphores, etc.).

Python's Global Interpreter Lock (GIL) means only one thread executes Python bytecode at a time, so CPU-bound tasks don't get true parallelism with threads.

Best suited for I/O-bound tasks (like reading files, network operations, waiting for user input) where threads can run while others wait.





-Multiprocessing

Involves multiple processes, each with its own separate memory space.

Processes don't share memory by default, so communication is done via inter-process communication (IPC) mechanisms like pipes or queues.

Because each process runs independently, multiprocessing can achieve true parallelism, making it great for CPU-bound tasks (like heavy computations).

More memory overhead than threads, because each process is separate.

11.What is memory management in Python?


Memory management in Python refers to the way Python allocates, tracks, and frees memory used by objects during program execution. It ensures efficient use of memory and helps avoid leaks or crashes.



Key aspects of Python's memory management:
1.Automatic Memory Allocation

When you create objects (like variables, lists, classes), Python automatically allocates the required memory.

You don't manually allocate or free memory like in languages such as C or C++.

2.Reference Counting

Python primarily uses reference counting to keep track of how many references (names, variables, containers) point to an object.

When the reference count of an object drops to zero (no references left), Python immediately frees the memory.

3.Garbage Collection (GC)

Reference counting can't handle circular references (objects referring to each other).

Python has a garbage collector to detect and clean up these cycles, running periodically to free memory that is no longer accessible.

4.Memory Pools (Python Memory Allocator)

Python uses its own memory allocator to manage small objects efficiently.

It allocates memory in blocks called pools to reduce fragmentation and improve speed.



-Why is memory management important?

Prevents memory leaks (unused objects lingering in memory).

Ensures efficient memory use for performance.

Simplifies programming by abstracting manual memory handling.

15.How does Python's garbage collection system work?

Python's garbage collection (GC) system is responsible for automatically reclaiming memory occupied by objects that are no longer in use, helping prevent memory leaks and keep your program running efficiently.



How Python's Garbage Collection Works
1.Reference Counting (Primary Mechanism)

Every object in Python keeps a reference count — how many references point to it.

When you create a new reference to an object, its count increases.

When a reference is deleted or goes out of scope, the count decreases.

Once the reference count drops to zero, Python immediately frees that object’s memory.



a = []       # reference count of list increases to 1
b = a        # reference count increases to 2
del a        # reference count decreases to 1
del b        # reference count drops to 0 -> object is deallocated


2.Handling Circular References (Cycle Detector)

Reference counting can’t clean up circular references, where objects reference each other, forming a cycle.

For example, two objects referencing each other but no longer referenced elsewhere will never have a ref count of zero.

To handle this, Python has a cyclic garbage collector in the gc module.

It periodically searches for groups of objects involved in cycles that are unreachable and frees them.






3.Generational Garbage Collection

Python's GC organizes objects into generations based on their lifespan:

Generation 0: Newly created objects.

Generation 1: Objects that survived one GC cycle.

Generation 2: Objects that survived multiple GC cycles.

The GC runs more frequently on younger generations (since most objects die young).

Older generations are collected less often to optimize performance.

23.What are the challenges associated with memory management in Python?

Memory management in Python is mostly automatic, thanks to features like reference counting and garbage collection, but it still comes with several challenges, especially for large or long-running applications.

Here are the most common challenges developers face:


1.Circular References

When two or more objects refer to each other, forming a cycle, reference counting alone can't free them.

Python's garbage collector handles this, but cycles can delay memory release.

In some cases, cycles involving objects with custom __del__() methods might not be collected at all.

Why it matters: Memory leaks can occur if cycles aren't cleaned up properly.


2.Memory Leaks

Even though Python has garbage collection, poor coding practices (like storing unused objects in global lists or caches) can prevent memory from being freed.

Objects held in global variables, long-lived containers, or closures may unintentionally persist.

Why it matters: Over time, memory leaks can cause performance degradation or even crashes.




3.Unpredictable Garbage Collection

Python doesn't guarantee exact timing of when memory will be released (unlike languages like C).

Objects with complex destruction logic might not be collected promptly.

Why it matters: Programs may temporarily consume more memory than expected.


4.Fragmentation

Python's memory allocator manages small objects efficiently using memory pools, but over time, fragmentation can occur.

Memory fragmentation can waste memory and reduce performance, especially in long-running applications.



5.Large Object Retention

Objects like large lists, NumPy arrays, or Pandas DataFrames may be unintentionally kept in memory.

Deleting variables (del) doesn't immediately release memory if other references exist.

Why it matters: High memory usage in data-heavy applications can become a bottleneck.


6.Difficulty Tracking Memory Usage

Python abstracts memory management, making it harder to know how much memory is being used or held by specific objects.

Tools like gc, tracemalloc, and objgraph help, but require effort to use effectively.


7.Global Interpreter Lock (GIL) Side Effects

Though not a memory issue directly, the GIL limits true multi-threaded memory parallelism.

As a result, developers may resort to multiprocessing, which increases memory overhead.

```python
"""4.Write a Python script that reads from one file and writes its content to another file."""

# Define the source and destination file names
source_file = 'source.txt'
destination_file = 'destination.txt'

try:
    # Open the source file in read mode
    with open(source_file, 'r') as src:
        # Read all contents
        content = src.read()

    # Open the destination file in write mode
    with open(destination_file, 'w') as dest:
        # Write the content to the destination file
        dest.write(content)

    print(f"Contents copied from '{source_file}' to '{destination_file}' successfully.")

except FileNotFoundError:
    print(f"Error: The file '{source_file}' does not exist.")
except IOError as e:
    print(f"An I/O error occurred: {e}")

```

```python
"""6.Write a Python program that logs an error message to a log file when a division by zero exception occurs."""

import logging

# Configure logging
logging.basicConfig(
    filename='error.log',           # Log file name
    level=logging.ERROR,            # Log only errors and above
    format='%(asctime)s - %(levelname)s - %(message)s'  # Log format
)

def divide(a, b):
    try:
        result = a / b
        print(f"Result: {result}")
    except ZeroDivisionError as e:
        logging.error("Attempted to divide by zero.")
        print("Error: Cannot divide by zero.")

# Example usage
divide(10, 0)

```

```python
"""7.How do you log information at different levels (INFO, ERROR, WARNING) in Python using the logging module."""

import logging

# Configure logging (prints to console)
logging.basicConfig(
    level=logging.DEBUG,  # Minimum level to log
    format='%(levelname)s:%(message)s'
)

logging.debug("This is a debug message")     # Very detailed info
logging.info("This is an info message")      # General info
logging.warning("This is a warning message") # Something suspicious
logging.error("This is an error message")    # Something went wrong
logging.critical("This is a critical message") # Very serious problem

```

```python
"""13.How would you check if a file exists before attempting to read it in Python."""

import os

filename = 'example.txt'

if os.path.exists(filename):
    with open(filename, 'r') as file:
        content = file.read()
        print(content)
else:
    print(f"The file '{filename}' does not exist.")

```

```python
"""14.Write a program that uses the logging module to log both informational and error messages."""

import logging

# Configure logging
logging.basicConfig(
    filename='app.log',               # Log file
    level=logging.INFO,               # Log INFO level and above
    format='%(asctime)s - %(levelname)s - %(message)s'  # Log format with timestamp
)

def divide_numbers(a, b):
    logging.info(f"Attempting to divide {a} by {b}")
    try:
        result = a / b
        logging.info(f"Division successful: {result}")
        return result
    except ZeroDivisionError:
        logging.error("Error: Division by zero attempted.")
        return None

# Example usage
divide_numbers(10, 2)
divide_numbers(10, 0)

```

```python
"""15.Write a Python program that prints the content of a file and handles the case when the file is empty."""

filename = 'example.txt'

try:
    with open(filename, 'r') as file:
        content = file.read()
        if content:
            print("File content:")
            print(content)
        else:
            print(f"The file '{filename}' is empty.")
except FileNotFoundError:
    print(f"Error: The file '{filename}' does not exist.")
except IOError:
    print(f"Error: Could not read the file '{filename}'.")

```

```python
"""18.How would you implement a basic logging setup that logs to a file with rotation after 1MB."""


import logging
from logging.handlers import RotatingFileHandler

# Create logger
logger = logging.getLogger('my_logger')
logger.setLevel(logging.DEBUG)  # Log all levels DEBUG and above

# Create a rotating file handler
handler = RotatingFileHandler(
    'app.log',       # Log file name
    maxBytes=1_000_000,  # Rotate after 1MB
    backupCount=3        # Keep last 3 log files as backup
)

# Create formatter and set it to handler
formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
handler.setFormatter(formatter)

# Add handler to logger
logger.addHandler(handler)

# Example usage
logger.info("This is an info message.")
logger.error("This is an error message.")

```

```python
"""21.Write a Python program that reads a file and prints the number of occurrences of a specific word."""


filename = 'example.txt'
word_to_count = 'python'

try:
    with open(filename, 'r') as file:
        content = file.read().lower()  # Read and convert to lowercase for case-insensitive count

    # Split content into words and count occurrences of the word
    words = content.split()
    count = words.count(word_to_count.lower())

    print(f"The word '{word_to_count}' occurs {count} times in the file.")

except FileNotFoundError:
    print(f"Error: The file '{filename}' does not exist.")
except IOError:
    print(f"Error: Could not read the file '{filename}'.")
```

```python
"""22.How can you check if a file is empty before attempting to read its contents."""


from pathlib import Path

filename = Path('example.txt')

if filename.exists():
    if filename.stat().st_size > 0:
        with filename.open('r') as file:
            content = file.read()
            print(content)
    else:
        print(f"The file '{filename}' is empty.")
else:
    print(f"The file '{filename}' does not exist.")
```

```python
"""23.Write a Python program that writes to a log file when an error occurs during file handling."""

import logging

# Configure logging to write errors to a file
logging.basicConfig(
    filename='file_errors.log',
    level=logging.ERROR,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

filename = 'example.txt'

try:
    with open(filename, 'r') as file:
        content = file.read()
        print(content)
except Exception as e:
    logging.error(f"An error occurred while handling the file '{filename}': {e}")
    print(f"An error occurred. Check 'file_errors.log' for details.")
```
