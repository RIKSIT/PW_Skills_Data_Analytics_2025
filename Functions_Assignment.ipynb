1. What is the difference between a function and a method in Python?





>>Function

-A function is a block of reusable code that performs a specific task.

-It is defined using the def keyword.

-Functions can exist independently (not tied to any object/class).

eg:
def sum(a,b):
    return a+b
sum(4,5)


>>Method

-A method is also a function, but it is associated with an object.

-Methods are defined inside a class and usually take self as their first parameter (referring to the object itself).

-Methods are called on objects using the dot (.) operator.

eg:
class name:
    def greet(self, name):
        return f"Hello, {name}!"

# Create an object
g = name()

print(g.greet("Rikshit"))

5. What are iterators in Python and how do they differ from iterables?





1. Iterable

An iterable is any Python object capable of returning its elements one at a time.

Examples: lists, tuples, strings, sets, dictionaries, ranges.

An iterable can be passed to the iter() function to get an iterator.

my_list = [1, 2, 3]

for x in my_list:   # list is iterable
    print(x)


2. Iterator

An iterator is an object that represents a stream of data.

It remembers its current position while iterating.

It has two important methods:

__iter__() → returns the iterator object itself.

__next__() → returns the next value, or raises StopIteration when no items are left.

my_list = [1, 2, 3]
it = iter(my_list)   # get iterator from iterable

print(next(it))   # 1
print(next(it))   # 2
print(next(it))   # 3
# print(next(it)) # would raise StopIteration




6. Explain the concept of generators in Python and how they are defined.





A generator is a special type of iterator that is created using a function with yield instead of return.

They allow you to produce a sequence of values lazily (one at a time, only when needed).

This saves memory and is useful for working with large data sets or infinite sequences.


def count_up_to(n):
    num = 1
    while num <= n:
        yield num
        num += 1

# Create generator
gen = count_up_to(5)

for value in gen:
    print(value)









7. What are the advantages of using generators over regular functions?






Advantages of Generators in Python
1. Memory Efficiency (Lazy Evaluation)

Generators don’t store all values in memory at once.

They produce values one at a time, only when needed.

Useful for large datasets or infinite sequences.

Example:

# Using a list comprehension (stores all 1M numbers in memory)
nums_list = [x*x for x in range(1_000_000)]

# Using a generator (creates values on demand)
nums_gen = (x*x for x in range(1_000_000))


Faster Execution for Large Data

Since they generate values on the fly, no need to wait for the entire sequence to be built.

Iteration starts immediately.

Example:
def big_numbers():
    for i in range(1_000_000):
        yield i

for num in big_numbers():
    if num == 5:
        break
print("Stopped at 5")

9. Explain the purpose and usage of the `map()` function in Python.





The map() function is used to apply a function to each item in an iterable (like a list, tuple, or string).

It returns a map object (iterator), which you usually convert to a list, tuple, or set.

Syntax:
map(function, iterable)

Example:
numbers = [1, 2, 3, 4]
result = map(lambda x: x * x, numbers)
print(list(result))   # [1, 4, 9, 16]






10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

1. map()

-Purpose: Applies a function to every element in an iterable.
-Output: Returns a new iterable (map object).

Example:
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, nums))
print(squares)   # [1, 4, 9, 16]


2. filter()

-Purpose: Filters elements from an iterable based on a condition (True/False).
-Output: Returns only those elements where the function returns True.

Example:
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)   # [2, 4, 6]


3. reduce()

-Purpose: Applies a function cumulatively to items in an iterable → reduces it to a single value.
-Output: Single value (sum, product, etc.).
-reduce() is in the functools module.

Example:

from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)   # 24 (1*2*3*4)

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(n):
  if n<=1:
    return False
  for i in range(2,int(n**0.5)+1):
    if n%i==0:
      return False
  return True

for num in range(1,201):
  print(f"{num} is {'Prime' if is_prime(num) else 'Not Prime'}")

```

```python
"""5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms."""

class FibonacciIterator:
    def __init__(self, n_terms):
        self.n_terms = n_terms   # total terms to generate
        self.count = 0           # how many terms generated so far
        self.a, self.b = 0, 1    # starting Fibonacci numbers

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n_terms:
            raise StopIteration  # stop iteration when limit is reached
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.b


# Example usage:
fib = FibonacciIterator(10)  # generate first 10 Fibonacci numbers
for num in fib:
    print(num, end=" ")

```

```python
#6. Write a generator function in Python that yields the powers of 2 up to a given exponent.


def power_of_two(n):
  for exp in range(n+1):
    yield 2**exp

for power in power_of_two(5):
  print(power)
```
