2.How does broadcasting work in NumPy?


Broadcasting in NumPy allows arithmetic operations between arrays of different shapes by automatically expanding them to a compatible shape—without copying data.

Two arrays are broadcast-compatible if:

They have the same number of dimensions, or one of them can be padded with 1s on the left.

For each dimension, the sizes must either be:

Equal, or

One of them is 1

example:
import numpy as np

a = np.array([1, 2, 3])
b = 5

print(a + b)  # Output: [6 7 8]

4.Explain the use of the groupby() method in Pandas.


The groupby() method in Pandas is used to split a DataFrame into groups, apply some function to each group (like sum, mean, etc.), and then combine the results into a new DataFrame or Series.

This is often referred to as the "Split-Apply-Combine" strategy:

Split the data into groups based on some criteria.

Apply a function to each group independently.

Combine the results into a new data structure.


example:
import pandas as pd

data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Finance'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Salary': [50000, 60000, 70000, 80000, 75000]
}

df = pd.DataFrame(data)

# Group by department and calculate average salary
grouped = df.groupby('Department')['Salary'].mean()

print(grouped)

Handling missing data ensures your DataFrame stays clean and ready for processing.

Enables smooth downstream processing like grouping, merging, or visualization.

17.Explain the difference between apply() and map() in Pandas.


1.apply()

Works on Series or DataFrames.

On a DataFrame, can specify axis=0 (apply function to each column) or axis=1 (apply function to each row).

Use when you want to run a function that processes rows or columns as a whole.


import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})

# Apply function on each column (sum)
print(df.apply(sum, axis=0))  # Output: A=6, B=15

# Apply function on each row (sum)
print(df.apply(sum, axis=1))  # Output: row-wise sums [5,7,9]


2.map()

Works only on Series.

Used to map or replace values in a Series according to a dictionary, Series, or function.

Best for element-wise transformations or value substitutions.

Example:
s = pd.Series(['cat', 'dog', 'rabbit'])

# Map strings to numbers
mapping = {'cat': 1, 'dog': 2}
print(s.map(mapping))
# Output: 1, 2, NaN (since 'rabbit' is unmapped)

21.Why is NumPy's array slicing faster than Python's list slicing?

1.Memory Layout and Data Type

NumPy arrays store elements in a contiguous block of memory with a fixed data type (e.g., all floats, ints).

Python lists are arrays of pointers to Python objects scattered in memory, often of mixed types.

2.View vs Copy

NumPy slicing returns a view of the original array — no data is copied.

Python list slicing returns a copy — a new list with copied elements.

Because NumPy doesn't copy data during slicing, it just creates a new array object pointing to the same memory, so slicing is very fast and memory-efficient.

3.Low-Level Implementation

NumPy is implemented in C, using optimized, compiled code for array operations.

Python lists operate at the Python interpreter level, which is slower due to dynamic type checking and overhead.

4.Contiguous Data and Strides

NumPy uses strides to efficiently access subsets of the data without copying.

Python lists don't have this concept — they must copy references during slicing.
