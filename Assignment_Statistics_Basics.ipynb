Question 2: What is sampling in statistics? Explain the differences between random and stratified sampling.



Sampling in statistics is the process of selecting a subset of individuals or items from a larger population to gather data and make inferences about the entire population. It's often impractical or impossible to collect data from every member of a population, so sampling allows us to study a smaller, representative group.

Here's the difference between random and stratified sampling:

Random Sampling: In simple random sampling, every member of the population has an equal chance of being selected for the sample. This helps to minimize bias and ensures that the sample is likely to be representative of the population. Imagine drawing names out of a hat - that's a form of random sampling.



Stratified Sampling: In stratified sampling, the population is divided into subgroups (strata) based on shared characteristics (e.g., age, gender, income). Then, a random sample is taken from each stratum. This method is useful when you want to ensure that certain subgroups are represented in your sample in proportion to their size in the population, or when you want to analyze differences between subgroups. For example, if you were studying student performance and wanted to ensure you had representation from different grade levels, you would stratify your sample by grade.
In summary, random sampling is about equal chance of selection, while stratified sampling involves dividing the population into groups and then randomly sampling from each group.





Question 3: Define mean, median, and mode. Explain why these measures of central
tendency are important.


Mean (Average): The mean is calculated by summing all the values in a dataset and dividing by the number of values.


Median: The median is the middle value in a dataset when the values are arranged in ascending or descending order. If there's an even number of values, the median is the average of the two middle values.



Mode: The mode is the value that appears most frequently in a dataset. A dataset can have one mode (unimodal), more than one mode (multimodal), or no mode.
Why are these measures of central tendency important?


These measures are important because they provide a single value that represents the typical or central value of a dataset. They help us:

Summarize data: They provide a concise way to describe the characteristics of a dataset.
Compare datasets: We can compare the central tendencies of different datasets to see how they differ.
Identify patterns: They can help reveal patterns and distributions within the data.
Make informed decisions: Understanding the central tendency can help in making decisions based on the data.
For example, if you're looking at the salaries of employees in a company, the mean salary gives you the average income, the median salary tells you the income of the middle-earning employee (less affected by extreme values), and the mode salary shows the most common income level. Each measure provides a different perspective on the typical salary in the company.






Question 4: Explain skewness and kurtosis. What does a positive skew imply about the data?


Skewness

Skewness is a measure of the asymmetry of the probability distribution of a real-valued random variable about its mean. In simpler terms, it tells you how much a dataset is "leaning" to one side.

Zero Skewness: The data is perfectly symmetrical, like a normal distribution (bell curve). The mean, median, and mode are all in the same place.
Positive Skew (Right Skew): The tail of the distribution extends to the right. This means there are more values concentrated on the left side of the distribution, and fewer, but potentially larger, values on the right.
What a positive skew implies: A positive skew indicates that the mean is typically greater than the median and mode. This often happens when there are a few unusually high values in the dataset that pull the mean towards the right. For example, if you look at the distribution of incomes in a country, it's often positively skewed because a small number of people earn very high incomes, pulling the average income up.
Negative Skew (Left Skew): The tail of the distribution extends to the left. This means there are more values concentrated on the right side of the distribution, and fewer, potentially smaller, values on the left.
Kurtosis

Kurtosis is a measure of the "tailedness" of the probability distribution of a real-valued random variable. It describes the shape of the distribution's peaks and tails relative to a normal distribution.

Mesokurtic: The distribution has a kurtosis similar to that of a normal distribution.
Leptokurtic: The distribution has heavier tails and a sharper peak than a normal distribution. This means there are more outliers (extreme values) in the data.
Platykurtic: The distribution has lighter tails and a flatter peak than a normal distribution. This means there are fewer outliers in the data.



```python
""": Implement a Python program to compute the mean, median, and mode of
a given list of numbers.
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]"""

import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

# Calculate the mean
mean_value = statistics.mean(numbers)
print(f"Mean: {mean_value}")

# Calculate the median
median_value = statistics.median(numbers)
print(f"Median: {median_value}")

# Calculate the mode
try:
    mode_value = statistics.mode(numbers)
    print(f"Mode: {mode_value}")
except statistics.StatisticsError:
    print("No unique mode found")
```

```python
"""
Question 6: Compute the covariance and correlation coefficient between the following
two datasets provided as lists in Python:
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

"""

import numpy as np

# given lists
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# convert to numpy arrays
x = np.array(list_x)
y = np.array(list_y)

# calculate covariance matrix (sample by default)
cov_matrix = np.cov(x, y, bias=True)   # bias=True → population covariance
cov_xy = cov_matrix[0, 1]

# calculate correlation coefficient
corr_matrix = np.corrcoef(x, y)
corr_xy = corr_matrix[0, 1]

print("Covariance:", cov_xy)
print("Correlation Coefficient:", corr_xy)
```

```python
"""
Question 7: Write a Python script to draw a boxplot for the following numeric list and
identify its outliers. Explain the result:
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
"""
import matplotlib.pyplot as plt
import numpy as np

# given data
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Draw boxplot
plt.boxplot(data, vert=False, patch_artist=True, boxprops=dict(facecolor="lightblue"))
plt.title("Boxplot of Data")
plt.xlabel("Values")
plt.show()

# Outlier detection using IQR
Q1 = np.percentile(data, 25)   # First quartile (25th percentile)
Q3 = np.percentile(data, 75)   # Third quartile (75th percentile)
IQR = Q3 - Q1                  # Interquartile range

# Define lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

1. How Covariance and Correlation Help

Covariance:

Measures how two variables move together.

If positive → when advertising spend increases, sales also tend to increase.

If negative → higher advertising spend is associated with lower sales.

But covariance is scale-dependent, so it's hard to interpret directly.

Correlation Coefficient (r):

Standardized version of covariance.

Always between -1 and +1.

r≈+1 → strong positive linear relationship.

r≈-1 → strong negative linear relationship.

r≈0 → little or no linear relationship.

Much easier to interpret than raw covariance.
correlation will show how strongly advertising spend is related to sales growth.

```python
#Python Code for ques 8
import numpy as np

# Data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Convert to numpy arrays
x = np.array(advertising_spend)
y = np.array(daily_sales)

# Covariance
cov_matrix = np.cov(x, y, bias=True)   # bias=True → population covariance
cov_xy = cov_matrix[0, 1]

# Correlation
corr_matrix = np.corrcoef(x, y)
corr_xy = corr_matrix[0, 1]

print("Covariance:", cov_xy)
print("Correlation Coefficient:", corr_xy)

```

Question 9


Your team has collected customer satisfaction survey data on a scale of
1-10 and wants to understand its distribution before launching a new product.
● Explain which summary statistics and visualizations (e.g. mean, standard
deviation, histogram) you'd use.

Summary Statistics

Mean (average) → overall central tendency.

Median → robust middle score, useful if skew/outliers exist.

Mode → most common score (important in survey data).

Standard deviation / Variance → spread of satisfaction levels.

Min & Max → range of responses.

Visualizations

Histogram → shows frequency distribution of satisfaction scores.

Boxplot → highlights median, quartiles, and outliers.

Bar chart (for categorical counts) → useful if treating each score (1-10) as category.

This combination helps the team quickly see if customers are generally satisfied (scores closer to 8-10) or dissatisfied (scores near 1-4), and whether responses are consistent or spread out.


```python
#Write Python code to create a histogram using Matplotlib for the survey data:
#survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

import matplotlib.pyplot as plt
import numpy as np

# Survey data
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

# Compute summary statistics
mean_score = np.mean(survey_scores)
median_score = np.median(survey_scores)
std_dev = np.std(survey_scores)

print("Mean:", mean_score)
print("Median:", median_score)
print("Standard Deviation:", std_dev)

# Create histogram
plt.hist(survey_scores, bins=range(1, 12), edgecolor="black", color="skyblue", rwidth=0.8)
plt.title("Customer Satisfaction Survey Distribution")
plt.xlabel("Survey Score (1-10)")
plt.ylabel("Frequency")
plt.xticks(range(1, 11))
plt.show()
```
